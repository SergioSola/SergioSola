# MFI-P1: Fiscal Dominance and the Policy Mix

## A DSGE Model of Monetary-Fiscal Interaction with Banking, Sovereign Risk, and Open Economy

**Author:** Sergio Sola | February 2026

---

### Abstract

This paper develops a medium-scale DSGE model to study how government debt limits
constrain central bank autonomy and shape macroeconomic dynamics. The model features
five types of agents — households, firms, banks, a government, and a central bank —
embedded in an open economy with foreign investors. Banks lend to both the government
and the private sector; when the government approaches its debt limit, the central bank
may become *captive*, keeping rates lower than prescribed by the Taylor rule. This
accommodation reduces default risk premia but raises inflation risk premia, with an
ambiguous net effect on sovereign spreads. Corporate borrowing costs, which carry a
spread over the sovereign rate, transmit fiscal stress to private credit, investment, and
firm default rates. We simulate the model under three policy regimes — (i) active monetary /
passive fiscal, (ii) passive monetary / active fiscal, and (iii) active monetary / active fiscal —
to characterise the macroeconomic consequences of the policy mix.

---

## 1. Introduction & Motivation

Public debt-to-GDP ratios have risen sharply across advanced economies since the
Global Financial Crisis, and further during the COVID-19 pandemic. In several countries,
debt levels now exceed historical norms, raising concerns about:

- **Fiscal sustainability:** Can primary surpluses stabilise debt at current levels?
- **Central bank independence:** Will inflation-targeting central banks face pressure
  to keep rates low to ensure debt sustainability?
- **Financial stability:** How does sovereign stress propagate through the banking
  sector to private credit conditions?
- **External balance:** How do foreign investors respond to deteriorating fiscal
  credibility, and what are the exchange rate implications?

The Sargent & Wallace (1981) "unpleasant monetarist arithmetic" established that
fiscal policy can dominate monetary policy when the government's intertemporal budget
constraint binds. More recently, the Fiscal Theory of the Price Level (Leeper 1991;
Cochrane 2023) has formalised how the active/passive classification of policy regimes
determines equilibrium outcomes.

This paper contributes by:
1. Endogenising the **transition to fiscal dominance** via a debt limit mechanism
2. Modelling the **risk premia trade-off** (inflation vs. default) explicitly
3. Embedding a **banking sector** that transmits sovereign stress to corporate credit
4. Adding an **open-economy dimension** with heterogeneous foreign investors

### 2.2 Households — Full Derivation

A representative household maximises lifetime utility:

$$\max_{\{C_t, N_t, D_t\}} \; E_0 \sum_{t=0}^{\infty} \beta^t \left[ \frac{(C_t - h C_{t-1})^{1-\sigma}}{1-\sigma} - \chi \frac{N_t^{1+\varphi}}{1+\varphi} \right]$$

subject to the sequential budget constraint:

$$C_t + D_t = W_t N_t + (1 + r_t^d) D_{t-1} + \Pi_t^{firms} + \Pi_t^{banks} - T_t$$

where $C_t$ is consumption, $N_t$ is labour, $D_t$ are bank deposits, $W_t$ is the real
wage, $r_t^d$ is the deposit rate, $\Pi_t$ are profit transfers, and $T_t$ are lump-sum taxes.
The parameter $h \in [0,1)$ governs external habit formation.

**Lagrangian.** Attach multiplier $\lambda_t$ to the budget constraint:

$$\mathcal{L} = E_0 \sum_{t=0}^{\infty} \beta^t \Bigg\{ \frac{(C_t - h C_{t-1})^{1-\sigma}}{1-\sigma} - \chi \frac{N_t^{1+\varphi}}{1+\varphi} + \lambda_t \Big[ W_t N_t + (1+r_t^d) D_{t-1} + \Pi_t - T_t - C_t - D_t \Big] \Bigg\}$$

**FOC w.r.t. $C_t$:**

$$\frac{\partial \mathcal{L}}{\partial C_t} = 0 \implies (C_t - h C_{t-1})^{-\sigma} - \beta h \, E_t \left[(C_{t+1} - h C_t)^{-\sigma}\right] = \lambda_t$$

Define the **effective marginal utility of consumption**:

$$\boxed{\Lambda_t \equiv (C_t - h C_{t-1})^{-\sigma} - \beta h \, E_t\left[(C_{t+1} - h C_t)^{-\sigma}\right]}$$

so that $\Lambda_t = \lambda_t$.

**FOC w.r.t. $N_t$:**

$$\frac{\partial \mathcal{L}}{\partial N_t} = 0 \implies \chi N_t^{\varphi} = \lambda_t W_t$$

This gives the **intratemporal labour supply condition**:

$$\boxed{\frac{\chi N_t^{\varphi}}{\Lambda_t} = W_t}$$

**FOC w.r.t. $D_t$:**

$$\frac{\partial \mathcal{L}}{\partial D_t} = 0 \implies \lambda_t = \beta \, E_t\left[\lambda_{t+1}(1 + r_{t+1}^d)\right]$$

This gives the **Euler equation** (consumption-savings):

$$\boxed{1 = \beta \, E_t\left[\frac{\Lambda_{t+1}}{\Lambda_t}(1 + r_{t+1}^d)\right]}$$

In the absence of arbitrage, the deposit rate equals the risk-free rate set by the
central bank: $r_t^d = i_t - E_t[\pi_{t+1}]$, where $i_t$ is the nominal policy rate
and $\pi_t$ is CPI inflation. Substituting:

$$1 = \beta \, E_t\left[\frac{\Lambda_{t+1}}{\Lambda_t} \cdot \frac{1 + i_t}{1 + \pi_{t+1}}\right]$$

**Stochastic discount factor.** The household's SDF, which prices all assets, is:

$$M_{t,t+1} = \beta \frac{\Lambda_{t+1}}{\Lambda_t} \cdot \frac{1}{1 + \pi_{t+1}}$$

### 2.3 Firms — Full Derivation

#### 2.3.1 Production Technology

A continuum of monopolistically competitive intermediate goods firms indexed by
$j \in [0,1]$ produce differentiated varieties using a Cobb-Douglas technology:

$$Y_t(j) = A_t \, K_t(j)^\alpha \, N_t(j)^{1-\alpha}$$

where $A_t$ is aggregate TFP (common to all firms), $K_t(j)$ is capital, and $N_t(j)$ is labour.
TFP follows the AR(1) process:

$$\log A_t = \rho_a \log A_{t-1} + \varepsilon_t^a, \qquad \varepsilon_t^a \sim N(0, \sigma_a^2)$$

Aggregate output is a CES composite:

$$Y_t = \left[\int_0^1 Y_t(j)^{\frac{\varepsilon_p - 1}{\varepsilon_p}} dj \right]^{\frac{\varepsilon_p}{\varepsilon_p - 1}}$$

where $\varepsilon_p > 1$ is the elasticity of substitution. The associated demand curve is:

$$Y_t(j) = \left(\frac{P_t(j)}{P_t}\right)^{-\varepsilon_p} Y_t$$

#### 2.3.2 Cost Minimisation

Each firm minimises cost $W_t N_t(j) + R_t^K K_t(j)$ subject to the production function.
The Lagrangian is:

$$\min_{N_t(j), K_t(j)} \; W_t N_t(j) + R_t^K K_t(j) + MC_t(j)\left[Y_t(j) - A_t K_t(j)^\alpha N_t(j)^{1-\alpha}\right]$$

**FOC w.r.t. $N_t(j)$:**

$$W_t = MC_t(j) \cdot (1-\alpha) A_t K_t(j)^\alpha N_t(j)^{-\alpha} = MC_t(j) \cdot (1-\alpha) \frac{Y_t(j)}{N_t(j)}$$

**FOC w.r.t. $K_t(j)$:**

$$R_t^K = MC_t(j) \cdot \alpha A_t K_t(j)^{\alpha-1} N_t(j)^{1-\alpha} = MC_t(j) \cdot \alpha \frac{Y_t(j)}{K_t(j)}$$

Dividing:

$$\frac{K_t(j)}{N_t(j)} = \frac{\alpha}{1-\alpha} \cdot \frac{W_t}{R_t^K}$$

This is identical across firms, so in symmetric equilibrium:

$$\boxed{MC_t = \frac{1}{A_t} \left(\frac{W_t}{1-\alpha}\right)^{1-\alpha} \left(\frac{R_t^K}{\alpha}\right)^{\alpha}}$$

In steady state, the real marginal cost equals the inverse markup: $MC = (\varepsilon_p - 1)/\varepsilon_p$.

#### 2.3.3 Calvo Pricing and the NKPC

Each period, a fraction $1-\theta_p$ of firms can re-optimise their price.
Non-optimising firms keep their price unchanged (zero indexation baseline).

A firm that re-optimises at $t$ chooses $P_t^*$ to maximise the expected discounted
sum of profits over the (random) duration for which the price remains in effect:

$$\max_{P_t^*} \; E_t \sum_{k=0}^{\infty} (\beta \theta_p)^k \frac{\Lambda_{t+k}}{\Lambda_t} \left[\frac{P_t^*}{P_{t+k}} - MC_{t+k}\right] Y_{t+k|t}$$

where $Y_{t+k|t} = (P_t^*/P_{t+k})^{-\varepsilon_p} Y_{t+k}$ is demand conditional on the price set at $t$.

**FOC** (taking derivative w.r.t. $P_t^*$ and simplifying):

$$\sum_{k=0}^{\infty} (\beta \theta_p)^k E_t\left[\Lambda_{t+k} Y_{t+k|t}\left(\frac{P_t^*}{P_{t+k}} - \frac{\varepsilon_p}{\varepsilon_p - 1} MC_{t+k}\right)\right] = 0$$

In the zero-inflation steady state, $P_t^* = P_t$ and $MC = (\varepsilon_p - 1)/\varepsilon_p$.

**Log-linearisation** of the pricing FOC and the price index $P_t = [\theta_p P_{t-1}^{1-\varepsilon_p} + (1-\theta_p)(P_t^*)^{1-\varepsilon_p}]^{1/(1-\varepsilon_p)}$ yields the standard **New Keynesian Phillips Curve**:

$$\boxed{\hat{\pi}_t = \beta \, E_t[\hat{\pi}_{t+1}] + \kappa \, \widehat{mc}_t}$$

where the **slope** is:

$$\kappa \equiv \frac{(1-\theta_p)(1-\beta\theta_p)}{\theta_p}$$

#### 2.3.4 Working Capital and BGG-Style Default

Firms finance a fraction of production costs through bank loans at rate $R_t^{corp}$.
Each firm draws an idiosyncratic productivity shock $\omega_t(j) \sim \text{LogNormal}\left(-\tfrac{1}{2}\sigma_\omega^2, \sigma_\omega^2\right)$
so that effective output is $\omega_t(j) \cdot Y_t(j)$, with $E[\omega] = 1$.

A firm **defaults** when revenue falls short of debt service:

$$\omega_t(j) \cdot Y_t(j) < R_t^{corp} \cdot L_t(j)$$

The **aggregate default probability** (Bernanke, Gertler & Gilchrist 1999):

$$\boxed{P_t^{def} = \Phi\!\left(\frac{\log\!\left(\frac{R_t^{corp} L_t}{Y_t}\right) + \tfrac{1}{2}\sigma_\omega^2}{\sigma_\omega}\right)}$$

where $\Phi(\cdot)$ is the standard normal CDF. Default probability rises with the
corporate lending rate and leverage, and falls with output.

### 2.4 Banks — Full Derivation

Banks intermediate between depositors (households) and borrowers (government, firms).

#### Balance Sheet

$$B_t^{gov} + L_t^{corp} = D_t + E_t^{bank}$$

where $B_t^{gov}$ are sovereign bond holdings, $L_t^{corp}$ are corporate loans,
$D_t$ are household deposits, and $E_t^{bank}$ is bank equity (net worth).

#### Capital Adequacy Constraint

Banks must satisfy a regulatory capital requirement:

$$E_t^{bank} \geq \kappa \left(B_t^{gov} + L_t^{corp}\right)$$

where $\kappa$ is the minimum capital-to-asset ratio (e.g. Basel III: $\kappa = 0.08$).

#### Bank Profit Maximisation

Banks choose their portfolio $(B_t^{gov}, L_t^{corp})$ to maximise profits:

$$\max \; R_t^{gov} B_t^{gov} + R_t^{corp}(1 - P_t^{def}) L_t^{corp} - r_t^d D_t$$

subject to the balance sheet identity and the capital constraint.

**FOC** (assuming the capital constraint binds):

The **corporate lending rate** carries a spread over the sovereign rate:

$$\boxed{R_t^{corp} = R_t^{gov} + \mu_t^{credit}}$$

The **credit spread** $\mu_t^{credit}$ compensates for:
1. Expected default losses: $P_t^{def}$
2. Capital scarcity: tighter capital → higher required return
3. Sovereign-bank nexus: losses on $B_t^{gov}$ erode $E_t^{bank}$, forcing deleveraging

We model the credit spread as:

$$\boxed{\mu_t^{credit} = \bar{\mu} + \zeta_1 \hat{P}_t^{def} + \zeta_2 \hat{b}_t}$$

where $\hat{P}_t^{def}$ and $\hat{b}_t$ are log-deviations from steady state. The term
$\zeta_2 \hat{b}_t$ captures the **sovereign-bank doom loop**: rising government debt
reduces the market value of bank sovereign holdings, eroding capital and tightening credit.

#### Bank Equity Dynamics

$$E_{t+1}^{bank} = E_t^{bank} + R_t^{gov} B_t^{gov} + R_t^{corp}(1-P_t^{def})L_t^{corp} - r_t^d D_t - \text{div}_t$$

where $\text{div}_t$ are dividends paid to households. In steady state, bank equity grows at
rate zero (dividends absorb all retained earnings).

### 2.5 Government — Full Derivation

#### Flow Budget Constraint

The government finances spending $G_t$ and debt service through taxes $T_t$ and new
bond issuance. Using end-of-period stock convention:

$$B_t = (1 + R_{t-1}^{gov}) B_{t-1} + G_t - T_t$$

Dividing by output $Y_t$ and defining $b_t \equiv B_t/Y_t$:

$$b_t = \frac{(1 + R_{t-1}^{gov})}{1 + g_t^Y} \, b_{t-1} + \frac{G_t - T_t}{Y_t}$$

where $g_t^Y$ is output growth. The **primary surplus** is $s_t \equiv (T_t - G_t)/Y_t$.

#### Intertemporal Budget Constraint

Iterating forward and imposing the no-Ponzi transversality condition:

$$b_t = \sum_{k=1}^{\infty} \left(\prod_{j=0}^{k-1} \frac{1}{1+r_{t+j}^{gov} - g_{t+j}^Y}\right) s_{t+k}$$

Debt is sustainable if and only if the present value of future primary surpluses
equals the current debt stock. This fails when $b_t$ approaches the **fiscal limit**.

#### Fiscal Rule

Tax revenues respond to debt and spending deviations:

$$\boxed{\tau_t = \bar{\tau} + \phi_b(b_t - \bar{b}) + \phi_g(g_t - \bar{g}) + \varepsilon_t^{fiscal}}$$

where $\tau_t = T_t/Y_t$, $g_t = G_t/Y_t$, and $\varepsilon_t^{fiscal}$ is a fiscal policy shock.

- **Passive fiscal policy** ($\phi_b$ large): taxes rise sufficiently with debt to ensure
  stability. The PVBC is satisfied for any bounded path of the price level.
- **Active fiscal policy** ($\phi_b$ small): taxes respond weakly to debt. The PVBC can
  only be satisfied if the price level adjusts (FTPL mechanism) or the CB accommodates.

#### Government Spending Process

$$\log(G_t/\bar{G}) = \rho_g \log(G_{t-1}/\bar{G}) + \varepsilon_t^g, \qquad \varepsilon_t^g \sim N(0, \sigma_g^2)$$

#### Fiscal Limit (Bi 2012)

There exists a maximum debt-to-GDP ratio $\bar{b}^{max}$ beyond which the required
fiscal adjustment exceeds what is politically feasible. For $b_t > \bar{b}^{max}$, at
least one of the following must occur: (i) sovereign default, (ii) inflation (monetary
accommodation), or (iii) external adjustment (currency crisis).

### 2.6 Central Bank — Full Derivation

#### Standard Taylor Rule (Active Monetary Policy)

$$\boxed{i_t = \rho_i \, i_{t-1} + (1 - \rho_i)\left[\bar{r} + \bar{\pi} + \phi_\pi(\pi_t - \bar{\pi}) + \phi_y \tilde{y}_t\right] + \varepsilon_t^{mp}}$$

where $\bar{r} = 1/\beta - 1$ is the natural rate, $\bar{\pi}$ is the inflation target,
$\tilde{y}_t = \hat{y}_t$ is the output gap, and $\varepsilon_t^{mp} \sim N(0, \sigma_{mp}^2)$.

The **Taylor principle** requires $\phi_\pi > 1$: the nominal rate rises more than
one-for-one with inflation, ensuring the real rate increases and demand is stabilised.

#### Captive Central Bank (Passive Monetary Policy)

When government debt approaches the fiscal limit, political pressure or fear of
sovereign default induces the CB to **accommodate** — keeping rates lower than the
Taylor rule prescribes. We model this as a **smooth logistic transition**:

$$\boxed{\phi_\pi^{eff}(b_t) = \phi_\pi - \frac{\phi_\pi - \phi_\pi^{cap}}{1 + \exp\!\left(-\gamma(b_t - \bar{b}^{max})\right)}}$$

**Properties:**
- When $b_t \ll \bar{b}^{max}$: $\phi_\pi^{eff} \approx \phi_\pi > 1$ (active, Taylor principle holds)
- When $b_t \approx \bar{b}^{max}$: $\phi_\pi^{eff} \approx (\phi_\pi + \phi_\pi^{cap})/2$ (transition)
- When $b_t \gg \bar{b}^{max}$: $\phi_\pi^{eff} \approx \phi_\pi^{cap} < 1$ (captive, Taylor principle violated)

The **degree of captivity** at any point is:

$$\mathcal{C}_t \equiv 1 - \frac{\phi_\pi^{eff}(b_t)}{\phi_\pi} \in [0, 1]$$

$\mathcal{C}_t = 0$ means full independence; $\mathcal{C}_t = 1$ means complete captivity.

### 2.7 Sovereign Risk Premia — Full Derivation

The sovereign bond yield carries two endogenous risk premia above the policy rate:

$$\boxed{R_t^{gov} = i_t + \psi_t^\pi + \psi_t^d}$$

#### Inflation Risk Premium

When the CB is captive, investors demand compensation for the risk that future
inflation will erode bond returns:

$$\boxed{\psi_t^\pi = \bar{\psi}^\pi \cdot \mathcal{C}_t^{\;\eta}}$$

where $\bar{\psi}^\pi$ is the maximum inflation premium (when fully captive) and
$\eta > 0$ controls the convexity. The premium is zero under active monetary policy
($\mathcal{C}_t = 0$) and rises nonlinearly with captivity.

#### Default Risk Premium

Investors also demand compensation for the probability that the government reneges on
its obligations. The default premium increases in debt but decreases in accommodation
(because an accommodating CB reduces outright default risk by inflating away debt):

$$\boxed{\psi_t^d = \bar{\psi}^d \cdot \exp\!\left(\delta_1(b_t - \bar{b}) - \delta_2 \, \mathcal{C}_t\right)}$$

- $\delta_1 > 0$: higher debt → higher default premium
- $\delta_2 > 0$: more accommodation → lower default premium (substitution effect)

#### The Ambiguous Net Effect

The key insight of the model: when the CB becomes captive,

$$\frac{dR^{gov}}{d\mathcal{C}} = \underbrace{\eta \bar{\psi}^\pi \mathcal{C}^{\eta-1}}_{>0 \text{ (inflation premium } \uparrow)} - \underbrace{\delta_2 \psi^d}_{>0 \text{ (default premium } \downarrow)}$$

Whether sovereign yields **rise or fall** when the CB accommodates depends on
whether $\eta\bar{\psi}^\pi\mathcal{C}^{\eta-1} \gtrless \delta_2\psi^d$.
This is **RQ2** of the paper — resolved through calibration and simulation.

### 2.8 Open Economy — Full Derivation

#### Uncovered Interest Parity with Risk Premium

Foreign investors can hold domestic sovereign bonds. The UIP condition (with a risk premium):

$$\boxed{E_t[e_{t+1}] - e_t = (i_t - i_t^*) + \xi_t}$$

where $e_t$ is the log nominal exchange rate (domestic per foreign), $i_t^*$ is
the foreign interest rate, and $\xi_t$ is the foreign risk premium.

#### Foreign Risk Premium

$$\boxed{\xi_t = \bar{\xi} + \xi_b(b_t - \bar{b}) + \xi_f(nfa_t - \overline{nfa}) + \varepsilon_t^\xi}$$

with $\varepsilon_t^\xi$ following AR(1): $\varepsilon_t^\xi = \rho_\xi \varepsilon_{t-1}^\xi + u_t^\xi$.

The premium rises with government debt (capital flight channel) and falls with
net foreign assets (external balance channel).

#### Real Exchange Rate

The real exchange rate is $Q_t = e_t + P_t^* - P_t$, or in log-deviations:

$$\hat{q}_t = \hat{e}_t + \hat{p}_t^* - \hat{p}_t$$

With constant foreign prices (small open economy assumption): $\hat{q}_t \approx \hat{e}_t - \hat{\pi}_t^{cum}$.

#### Trade Balance

Net exports respond to the real exchange rate:

$$NX_t / Y_t = \alpha_{open} \hat{q}_t$$

where $\alpha_{open}$ is the import share of the consumption basket. A real depreciation
($\hat{q}_t > 0$) improves the trade balance.

and policy/state variables $\{b_t, G_t, T_t, A_t, \xi_t\}$ such that:

### 3.2 Log-Linearisation

All variables are expressed as log-deviations from steady state:
$\hat{x}_t \equiv \log(X_t/\bar{X}) \approx (X_t - \bar{X})/\bar{X}$.

#### Equation 1: IS Curve (Euler equation with habit)

Starting from the Euler equation and log-linearising around $\bar{C}$, $\bar{\pi}$:

$$\boxed{\hat{c}_t = \frac{h}{1+h}\hat{c}_{t-1} + \frac{1}{1+h}E_t[\hat{c}_{t+1}] - \frac{1-h}{(1+h)\sigma}\left(\hat{\imath}_t - E_t[\hat{\pi}_{t+1}]\right)}$$

*Derivation:* The effective marginal utility $\Lambda_t = (C_t - hC_{t-1})^{-\sigma}$
(ignoring the $\beta h$ term at first order). Log-linearising:
$\hat{\Lambda}_t \approx -\sigma/(1-h)[\hat{c}_t - h/(1+h)\hat{c}_{t-1}]$.
Substituting into the Euler equation $\hat{\Lambda}_t = E_t[\hat{\Lambda}_{t+1}] + \hat{\imath}_t - E_t[\hat{\pi}_{t+1}]$
and rearranging gives the habit-adjusted IS curve above.

#### Equation 2: New Keynesian Phillips Curve

$$\boxed{\hat{\pi}_t = \beta E_t[\hat{\pi}_{t+1}] + \kappa\left(\hat{w}_t - \hat{a}_t + \alpha\hat{n}_t\right)}$$

where $\kappa = (1-\theta_p)(1-\beta\theta_p)/\theta_p$ and
$\widehat{mc}_t = \hat{w}_t - \hat{a}_t + \alpha\hat{n}_t$ (with capital fixed at SS).

*Derivation:* From the Calvo pricing FOC, the reset price $\hat{p}_t^* = (1-\beta\theta_p)\sum_{k=0}^\infty(\beta\theta_p)^k E_t[\widehat{mc}_{t+k} + \hat{p}_{t+k}]$.
From the price index: $\hat{\pi}_t = (1-\theta_p)(\hat{p}_t^* - \hat{p}_{t-1})$.
Combining and simplifying yields the recursive NKPC.

#### Equation 3: Labour Supply

$$\boxed{\hat{w}_t = \frac{\sigma}{1-h}\hat{c}_t + \varphi\hat{n}_t}$$

*Derivation:* Log-linearise $W = \chi N^\varphi / \Lambda$:
$\hat{w}_t = \varphi\hat{n}_t - \hat{\Lambda}_t = \varphi\hat{n}_t + \sigma/(1-h)\hat{c}_t$.

#### Equation 4: Production Function

$$\boxed{\hat{y}_t = \hat{a}_t + (1-\alpha)\hat{n}_t}$$

(Capital is predetermined at its steady-state level in the linearised system.)

#### Equation 5: Resource Constraint

$$\boxed{\hat{y}_t = \frac{C}{Y}\hat{c}_t + \frac{G}{Y}\hat{g}_t + \alpha_{open}\hat{q}_t}$$

(Investment is constant at $\delta K$ in the linearised system; net exports enter via $q$.)

#### Equation 6: Taylor Rule

$$\boxed{\hat{\imath}_t = (1-\rho_i)\left[\phi_\pi^{eff}\hat{\pi}_t + \phi_y\hat{y}_t\right] + \varepsilon_t^{mp}}$$

where $\phi_\pi^{eff} = \phi_\pi$ (AM regimes) or $\phi_\pi^{cap}$ (PM/AF regime).

#### Equation 7: Sovereign Yield

$$\boxed{\hat{R}_t^{gov} = \frac{i}{R^{gov}}\hat{\imath}_t + \frac{\psi^d}{R^{gov}}\delta_1\bar{b}\,\hat{b}_t + \text{(inflation premium terms)}}$$

*Derivation:* Linearise $R^{gov} = i + \psi^\pi + \psi^d$ using the chain rule.
The default premium enters through $\hat{\psi}_t^d \approx \delta_1\bar{b}\,\hat{b}_t$
(at the AM/PF steady state where captivity is zero).

#### Equation 8: Corporate Rate

$$\boxed{\hat{R}_t^{corp} = \frac{R^{gov}}{R^{corp}}\hat{R}_t^{gov} + \frac{\mu}{R^{corp}}\hat{\mu}_t}$$

#### Equation 9: Credit Spread

$$\boxed{\hat{\mu}_t = \zeta_1\hat{P}_t^{def} + \zeta_2\hat{b}_t}$$

(Already linear; $\zeta_1 = 0.5$, $\zeta_2 = \lambda_{bank} = 0.10$.)

#### Equation 10: Default Probability

$$\boxed{\hat{P}_t^{def} = \frac{\phi(z)}{\Phi(z)\sigma_\omega}\hat{R}_t^{corp} - \frac{\phi(z)}{\Phi(z)\sigma_\omega}\hat{y}_t}$$

where $z = (\log(R^{corp}L/Y) + \sigma_\omega^2/2)/\sigma_\omega$ evaluated at SS.

#### Equation 11: UIP (Real Exchange Rate)

$$\boxed{\hat{q}_t = E_t[\hat{q}_{t+1}] - (\hat{\imath}_t - E_t[\hat{\pi}_{t+1}]) + \hat{\xi}_t}$$

#### Equation 12: Debt Law of Motion

$$\boxed{\hat{b}_{t+1} = \left(\frac{1}{\beta} - \phi_b\right)\hat{b}_t + \frac{\bar{b}}{\beta}\hat{R}_t^{gov} + \left(\frac{G}{Y}-\phi_g\right)\hat{g}_t - \frac{\phi_b}{2}\hat{y}_t + \varepsilon_t^{fiscal}}$$

*Derivation:* Linearise the government budget constraint around SS, substituting the
fiscal rule for $\hat{\tau}_t$. The term $\phi_b/2 \cdot \hat{y}_t$ captures the automatic
stabiliser (tax revenue rises with output).

#### Equations 13–15: Exogenous Processes

$$\hat{a}_{t+1} = \rho_a\hat{a}_t + \varepsilon_t^a$$
$$\hat{g}_{t+1} = \rho_g\hat{g}_t + \varepsilon_t^g$$
$$\hat{\xi}_{t+1} = \rho_\xi\hat{\xi}_t + \varepsilon_t^\xi$$

### 3.3 State-Space Representation

Stack the system as $A\,E_t[z_{t+1}] = B\,z_t + C\,\varepsilon_t$ where:

$$z_t = \underbrace{[\hat{y}_t, \hat{c}_t, \hat{\pi}_t, \hat{n}_t, \hat{w}_t, \hat{\imath}_t, \hat{R}_t^{gov}, \hat{R}_t^{corp}, \hat{\mu}_t, \hat{P}_t^{def}, \hat{q}_t}_{\text{11 controls (forward-looking)}}, \underbrace{\hat{b}_t, \hat{a}_t, \hat{g}_t, \hat{\xi}_t]}_{\text{4 states (predetermined)}}$$

$$\varepsilon_t = [\varepsilon_t^{fiscal}, \varepsilon_t^{mp}, \varepsilon_t^a, \varepsilon_t^\xi]$$

The system is solved using the **generalised Schur (QZ) decomposition**.
The **Blanchard-Kahn condition** requires that the number of eigenvalues of $(A,B)$
with modulus $>1$ equals the number of forward-looking variables (11).
This is satisfied under AM/PF and PM/AF but may fail under AM/AF (explosive dynamics).

## 3. Policy Regimes

The three regimes to be simulated:

| Regime | Monetary Policy | Fiscal Policy | Description |
|--------|----------------|---------------|-------------|
| **(a) AM/PF** | Active ($\phi_\pi > 1$) | Passive ($\phi_b$ large) | Standard inflation targeting; fiscal authority stabilises debt |
| **(b) PM/AF** | Passive ($\phi_\pi < 1$) | Active ($\phi_b$ small) | Fiscal dominance; CB accommodates, price level adjusts |
| **(c) AM/AF** | Active ($\phi_\pi > 1$) | Active ($\phi_b$ small) | Policy conflict; no stable equilibrium without modification |

### Regime (a): Active Monetary / Passive Fiscal
- Taylor principle satisfied → monetary policy pins down inflation
- Fiscal rule stabilises debt → sovereign risk premia remain low
- Standard New Keynesian transmission mechanism

### Regime (b): Passive Monetary / Active Fiscal
- Central bank captive → inflation determined by fiscal needs
- Government debt is sustainable via seigniorage / inflation erosion
- Inflation risk premium rises, default risk premium may fall
- Higher corporate spreads despite lower policy rate

### Regime (c): Active Monetary / Active Fiscal
- Both authorities pursue independent objectives
- Monetary tightening raises debt service → debt spirals
- Fiscal expansion raises demand → more tightening needed
- Potentially explosive dynamics; may require regime switch or crisis
