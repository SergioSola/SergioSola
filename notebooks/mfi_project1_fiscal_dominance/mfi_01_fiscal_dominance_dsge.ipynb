### Abstract

This paper develops a medium-scale DSGE model to study how government debt limits
constrain central bank autonomy and shape macroeconomic dynamics. The model features
five types of agents — households, firms, banks, a government, and a central bank —
embedded in an open economy with foreign investors. Banks lend to both the government
and the private sector; when the government approaches its debt limit, the central bank
may become *captive*, keeping rates lower than prescribed by the Taylor rule. This
accommodation reduces default risk premia but raises inflation risk premia, with an
ambiguous net effect on sovereign spreads. Corporate borrowing costs, which carry a
spread over the sovereign rate, transmit fiscal stress to private credit, investment, and
firm default rates. We simulate the model under three policy regimes — (i) active monetary /
passive fiscal, (ii) passive monetary / active fiscal, and (iii) active monetary / active fiscal —
to characterise the macroeconomic consequences of the policy mix.

## 1. Introduction & Motivation

Public debt-to-GDP ratios have risen sharply across advanced economies since the
Global Financial Crisis, and further during the COVID-19 pandemic. In several countries,
debt levels now exceed historical norms, raising concerns about:

- **Fiscal sustainability:** Can primary surpluses stabilise debt at current levels?
- **Central bank independence:** Will inflation-targeting central banks face pressure
  to keep rates low to ensure debt sustainability?
- **Financial stability:** How does sovereign stress propagate through the banking
  sector to private credit conditions?
- **External balance:** How do foreign investors respond to deteriorating fiscal
  credibility, and what are the exchange rate implications?

The Sargent & Wallace (1981) "unpleasant monetarist arithmetic" established that
fiscal policy can dominate monetary policy when the government's intertemporal budget
constraint binds. More recently, the Fiscal Theory of the Price Level (Leeper 1991;
Cochrane 2023) has formalised how the active/passive classification of policy regimes
determines equilibrium outcomes.

This paper contributes by:
1. Endogenising the **transition to fiscal dominance** via a debt limit mechanism
2. Modelling the **risk premia trade-off** (inflation vs. default) explicitly
3. Embedding a **banking sector** that transmits sovereign stress to corporate credit
4. Adding an **open-economy dimension** with heterogeneous foreign investors

## 2. Model Structure

### 2.1 Agents and Timing

```
┌─────────────┐     labour, dividends      ┌──────────────┐
│  Households  │◄──────────────────────────►│    Firms     │
│  (own firms  │     consumption goods       │  (produce,   │
│   & banks)   │                             │   borrow)    │
└──────┬───────┘                             └──────┬───────┘
       │ deposits                                   │ corporate loans
       ▼                                            ▼
┌──────────────┐    sovereign bonds    ┌───────────────────┐
│    Banks     │◄─────────────────────►│    Government     │
│  (lend to    │    corporate loans    │  (tax, spend,     │
│   gov + pvt) │◄─────────────────────►│   issue debt)     │
└──────┬───────┘                       └────────┬──────────┘
       │                                        │
       │         ┌──────────────┐               │ debt limit
       │         │ Central Bank │               │ pressure

       │         │  ± captivity)│
       │         └──────────────┘
       │

│   Sector     │  invests in domestic sovereign bonds
│              │  balance of payments
└──────────────┘
```

### 2.2 Households

Representative household maximises:

$$E_0 \sum_{t=0}^{\infty} \beta^t \left[ \frac{(C_t - hC_{t-1})^{1-\sigma}}{1-\sigma} - \chi \frac{N_t^{1+\varphi}}{1+\varphi} \right]$$

Subject to budget constraint:

$$C_t + D_t = W_t N_t + (1 + r_t^d) D_{t-1} + \Pi_t^{firms} + \Pi_t^{banks} - T_t$$

where $h$ is habit persistence, $D_t$ are bank deposits, and $\Pi$ are dividend transfers.

### 2.3 Firms

Intermediate goods firms produce with:

$$Y_t = A_t K_t^\alpha N_t^{1-\alpha}$$

Firms need **working capital** financed through bank loans at rate $R_t^{corp}$.
They face idiosyncratic productivity shocks $\omega_t \sim \text{LogNormal}(-\frac{1}{2}\sigma_\omega^2, \sigma_\omega^2)$
and default when $\omega_t A_t K_t^\alpha N_t^{1-\alpha} < R_t^{corp} L_t$.

**Default probability** (Merton/BGG style):

$$P_t^{def} = \Phi\left( \frac{\log(R_t^{corp} L_t / Y_t) + \frac{1}{2}\sigma_\omega^2}{\sigma_\omega} \right)$$

Nominal rigidities via Calvo pricing: fraction $\theta_p$ of firms cannot re-optimise prices each period.

### 2.4 Banks

Bank balance sheet:

$$B_t^{gov} + L_t^{corp} = D_t + E_t^{bank}$$

where $B_t^{gov}$ is sovereign bond holdings, $L_t^{corp}$ is corporate lending,
$D_t$ is deposits, and $E_t^{bank}$ is bank equity/capital.

**Corporate loan pricing:**

$$R_t^{corp} = R_t^{gov} + \underbrace{\mu_t^{credit}}_{\text{credit spread}}$$

Credit spread depends on:
- Expected firm default rate $P_t^{def}$
- Bank capital ratio $E_t^{bank} / (B_t^{gov} + L_t^{corp})$
- Aggregate uncertainty

Banks face a **capital adequacy constraint**: $E_t^{bank} \geq \kappa (B_t^{gov} + L_t^{corp})$

### 2.5 Government

Government budget constraint:

$$B_t^{gov} = (1 + R_{t-1}^{gov}) B_{t-1}^{gov} + G_t - T_t$$

**Fiscal rule:**

$$\tau_t = \bar{\tau} + \phi_b (b_t - \bar{b}) + \phi_g (g_t - \bar{g}) + \varepsilon_t^{fiscal}$$

where $b_t = B_t^{gov}/Y_t$ is the debt-to-GDP ratio.

**Debt limit:** There exists $\bar{b}^{\max}$ such that for $b_t > \bar{b}^{\max}$,
the required fiscal adjustment exceeds political/economic feasibility.

**Active fiscal policy:** $\phi_b$ is small (insufficient response to debt) → debt is non-stationary absent monetary accommodation.

**Passive fiscal policy:** $\phi_b$ is large enough to stabilise debt.

### 2.6 Central Bank

**Standard Taylor rule (Active monetary policy):**

$$i_t = \rho_i i_{t-1} + (1-\rho_i) \left[ \bar{r} + \phi_\pi (\pi_t - \bar{\pi}) + \phi_y \tilde{y}_t \right] + \varepsilon_t^{mp}$$

with $\phi_\pi > 1$ (Taylor principle satisfied).

**Captive central bank (Passive monetary policy):**

When $b_t \to \bar{b}^{\max}$, the effective rule becomes:

$$i_t = \min\left\{ i_t^{Taylor},\; \bar{r} + \phi_\pi^{cap} (\pi_t - \bar{\pi}) \right\}$$

where $\phi_\pi^{cap} < 1$ — the central bank no longer satisfies the Taylor principle.

The transition can be modelled as a smooth logistic function:

$$\phi_\pi^{eff}(b_t) = \phi_\pi - \frac{\phi_\pi - \phi_\pi^{cap}}{1 + \exp(-\gamma(b_t - \bar{b}^{\max}))}$$

### 2.7 Sovereign Risk Premia

Sovereign bond yield:

$$R_t^{gov} = i_t + \psi_t^\pi + \psi_t^d$$

**Inflation risk premium** $\psi_t^\pi$: increasing in expected future monetary accommodation

$$\psi_t^\pi = \bar{\psi}^\pi \cdot \left(1 - \frac{\phi_\pi^{eff}(b_t)}{\phi_\pi}\right)^\eta$$

**Default risk premium** $\psi_t^d$: increasing in debt, decreasing in central bank accommodation

$$\psi_t^d = \bar{\psi}^d \cdot \exp\left(\delta_1 (b_t - \bar{b}) - \delta_2 (1 - \phi_\pi^{eff}/\phi_\pi)\right)$$

The **net effect** on $R_t^{gov}$ when captivity activates is ambiguous:
- $\psi_t^\pi$ rises (inflation risk increases)
- $\psi_t^d$ may fall (accommodation reduces outright default risk)
- Which dominates is a key empirical/calibration question (RQ2)

### 2.8 Open Economy

**Uncovered interest parity with risk premium:**

$$E_t[e_{t+1}] - e_t = (i_t - i_t^*) + \xi_t$$

where $e_t$ is the log nominal exchange rate (domestic per foreign).

**Foreign risk premium:**

$$\xi_t = \bar{\xi} + \xi_b (b_t - \bar{b}) + \xi_f (nfa_t - \overline{nfa}) + \varepsilon_t^{\xi}$$

The foreign risk premium $\xi_t$ may differ from the domestic inflation risk premium
(modelling **home bias** / segmentation friction).

**Balance of payments:**

$$CA_t + KA_t = 0$$

Foreign investors hold a share $f_t$ of sovereign bonds:

$$f_t = \bar{f} + \phi_f \left( R_t^{gov} + E_t[\Delta e_{t+1}] - i_t^* - \xi_t \right)$$

## 3. Policy Regimes

The three regimes to be simulated:

| Regime | Monetary Policy | Fiscal Policy | Description |
|--------|----------------|---------------|-------------|
| **(a) AM/PF** | Active ($\phi_\pi > 1$) | Passive ($\phi_b$ large) | Standard inflation targeting; fiscal authority stabilises debt |
| **(b) PM/AF** | Passive ($\phi_\pi < 1$) | Active ($\phi_b$ small) | Fiscal dominance; CB accommodates, price level adjusts |
| **(c) AM/AF** | Active ($\phi_\pi > 1$) | Active ($\phi_b$ small) | Policy conflict; no stable equilibrium without modification |

### Regime (a): Active Monetary / Passive Fiscal
- Taylor principle satisfied → monetary policy pins down inflation
- Fiscal rule stabilises debt → sovereign risk premia remain low
- Standard New Keynesian transmission mechanism

### Regime (b): Passive Monetary / Active Fiscal
- Central bank captive → inflation determined by fiscal needs
- Government debt is sustainable via seigniorage / inflation erosion
- Inflation risk premium rises, default risk premium may fall
- Higher corporate spreads despite lower policy rate

### Regime (c): Active Monetary / Active Fiscal
- Both authorities pursue independent objectives
- Monetary tightening raises debt service → debt spirals
- Fiscal expansion raises demand → more tightening needed
- Potentially explosive dynamics; may require regime switch or crisis
