**Author:** Sergio Sola | February 2026

## Unifying Theme

This programme investigates the **joint determination of monetary and fiscal policy**
and its consequences for macroeconomic stability, sovereign risk, and the financial sector.
The central insight is that monetary and fiscal authorities do not operate in isolation:
government debt dynamics constrain central bank autonomy, while interest-rate policy
feeds back onto fiscal sustainability, private credit conditions, and external balance sheets.

The programme employs **structural DSGE modelling** enriched with banking frictions,
sovereign debt limits, and open-economy features to study how the **policy mix**—the
combination of active/passive monetary and fiscal stances—shapes macroeconomic outcomes
under different shock scenarios.

## MFI-P1: Fiscal Dominance and the Policy Mix

### Motivation

In many advanced and emerging economies, public debt has risen to levels that may
constrain central bank independence. When governments approach their **debt limit**,
the central bank may become *captive*—forced to keep interest rates lower than the
inflation-targeting mandate would prescribe—to prevent a fiscal crisis. This creates a
fundamental tension:

- **Lower policy rates** reduce government borrowing costs but may fuel inflation.
- **Inflation risk premia** on sovereign debt may rise to compensate investors for
  the erosion of real returns, partially or fully offsetting the decline in default risk
  premia (if any).
- **Private credit conditions** deteriorate as corporate loan rates carry a spread over
  the sovereign rate: higher sovereign risk premia raise the floor for all borrowing costs.
- **Foreign investors** demand additional compensation for currency and country risk,
  potentially amplifying capital outflows when fiscal credibility weakens.

This project builds a **medium-scale DSGE model** that captures these feedback loops
and uses it to study the macroeconomic consequences of different policy regimes.

### Research Questions

1. **RQ1 — Fiscal dominance threshold:** At what debt-to-GDP level does the central
   bank effectively lose monetary autonomy? How does this threshold depend on the
   credibility of the fiscal rule and the structure of debt (maturity, currency, holder composition)?

2. **RQ2 — Risk premia decomposition:** When the central bank is captive, does the
   *inflation risk premium* on sovereign bonds dominate the reduction in the *default risk
   premium*? Under what conditions is the net effect on sovereign spreads positive vs. negative?

3. **RQ3 — Credit channel:** How does the sovereign-banking nexus transmit fiscal stress
   to private credit? What is the quantitative impact on corporate default probabilities,
   investment, and output?

4. **RQ4 — Policy regimes:** How do macroeconomic dynamics differ under:
   - (a) Active monetary / passive fiscal policy (standard inflation targeting)
   - (b) Passive monetary / active fiscal policy (fiscal dominance)
   - (c) Active monetary / active fiscal policy (policy conflict)

### Model Structure

#### Agents

| Agent | Role | Key Features |
|-------|------|--------------|
| **Households** | Own firms and banks, consume, save, supply labour | Standard preferences with habit formation |
| **Firms** | Produce intermediate/final goods, borrow from banks | Nominal rigidities (Calvo/Rotemberg), working capital needs, default risk |
| **Banks** | Intermediate funds, lend to government and firms | Balance sheet: deposits + equity → sovereign bonds + corporate loans |
| **Government** | Taxes, spends, issues debt | Fiscal rule with debt limit; when constrained, pressures CB |
| **Central Bank** | Sets policy rate via Taylor rule | Rule subject to fiscal dominance: rate capped when debt limit binds |
| **Foreign Sector** | Invests in domestic assets, trades goods | Balance of payments, interest rate parity with time-varying risk premium |

#### Key Mechanisms

1. **Sovereign debt limit:**
   - Government follows fiscal rule: $\tau_t = \bar{\tau} + \phi_b (b_t - \bar{b})$
   - Debt limit $\bar{b}^{\max}$ beyond which fiscal adjustment becomes insufficient
   - When $b_t \to \bar{b}^{\max}$, central bank captivity activates

2. **Central bank captivity:**
   - Standard Taylor rule: $i_t = \rho_i i_{t-1} + (1-\rho_i)[\bar{r} + \phi_\pi (\pi_t - \bar{\pi}) + \phi_y \tilde{y}_t]$
   - Under captivity: $i_t = \min\{i_t^{Taylor}, i_t^{fiscal\_cap}\}$
   - The fiscal cap ensures sovereign debt service remains sustainable

3. **Sovereign risk premia:**
   - Sovereign bond yield: $R_t^{gov} = i_t + \underbrace{\psi^{\pi}(\cdot)}_{\text{inflation risk premium}} + \underbrace{\psi^{d}(\cdot)}_{\text{default risk premium}}$
   - Inflation risk premium increases with expected future monetary accommodation
   - Default risk premium depends on distance to debt limit
   - **Open question:** net effect when captivity reduces default risk but raises inflation risk

4. **Banking sector:**
   - Banks hold sovereign bonds and corporate loans
   - Corporate loan rate: $R_t^{corp} = R_t^{gov} + \text{spread}_t$
   - Spread depends on firm leverage, expected default rate, bank capital adequacy
   - Sovereign stress → higher floor for corporate rates → credit contraction

5. **Corporate default:**
   - Firms face idiosyncratic productivity shocks (Bernanke-Gertler-Gilchrist style)
   - Default probability: $P_t^{def} = \Phi\left(\frac{\log(D_t/A_t) + (R_t^{corp} - \mu_t + \frac{1}{2}\sigma^2)T}{\sigma\sqrt{T}}\right)$
   - Higher borrowing costs → more defaults → bank losses → further credit tightening

6. **Open economy:**
   - Uncovered interest parity: $E_t[\Delta e_{t+1}] = i_t - i_t^* + \xi_t$
   - Time-varying risk premium $\xi_t$ for foreign investors, potentially different from domestic
   - Models home bias / segmentation friction
   - Balance of payments: current account + capital account = 0
   - Foreign demand for domestic sovereign bonds responds to yield differential and risk premium

### Empirical Strategy

1. **Calibration:** Match steady-state ratios to a representative advanced economy
   (or panel of advanced economies with high debt)
2. **Estimation:** Bayesian estimation of key structural parameters using
   macro time series (output, inflation, interest rates, debt, exchange rate)
3. **Policy experiments:** Simulate regime-specific impulse responses:
   - Fiscal shock (spending increase, tax cut)
   - Monetary shock (unexpected tightening)
   - Productivity shock
   - Foreign risk premium shock
   - Debt limit shock (sudden loss of fiscal space)
4. **Regime comparison:** Welfare analysis across the three policy configurations

### Literature Positioning

- **Fiscal theory of the price level:** Leeper (1991), Sims (1994, 2011), Cochrane (2023)
  — Active vs. passive fiscal/monetary regimes; price-level determination when fiscal
  policy is active.
- **Sovereign-bank nexus:** Brunnermeier et al. (2016) "diabolic loop"; Farhi & Tirole (2018)
  — Banks hold sovereign bonds → sovereign stress feeds into bank stress and vice versa.
- **Fiscal limits and risk premia:** Bi (2012), Bi et al. (2013) — Endogenous fiscal limits
  and their impact on sovereign spreads.
- **Financial frictions in DSGE:** Bernanke, Gertler & Gilchrist (1999); Gertler & Karadi (2011)
  — Financial accelerator and banking sector frictions in New Keynesian models.
- **Open economy with sovereign risk:** Uribe & Schmitt-Grohé (2017); Corsetti et al. (2013)
  — Small open economy models with sovereign default and monetary policy.
- **Fiscal dominance:** Sargent & Wallace (1981) "unpleasant monetarist arithmetic";
  Bianchi & Melosi (2019) — Regime switching between monetary and fiscal dominance.

### Identification & Robustness

- **Parameter identification:** Dynare identification toolbox; sensitivity analysis
  for weakly identified parameters.
- **Regime robustness:** Compare smooth transition (logistic) vs. discrete switching
  for the fiscal dominance threshold.
- **Risk premia decomposition:** Test alternative specifications for how inflation
  risk premium and default risk premium interact (additive, multiplicative, regime-dependent).
- **Open question on premia dominance:** Formally test whether the inflation risk premium
  dominates the decline in default risk premium under captivity — this is an empirical
  question that the model can help discipline.

### Expected Contributions

1. A unified DSGE framework with **endogenous fiscal dominance** that activates
   when debt approaches its limit, rather than being imposed exogenously.
2. Quantitative assessment of the **risk premia trade-off**: under what parameter
   configurations does inflation risk premium dominate reduced default risk premium.
3. Characterisation of the **credit channel** of fiscal stress through the
   sovereign-bank-corporate nexus.
4. Policy guidance on the **welfare costs** of different monetary-fiscal regimes,
   including the often-discussed but rarely modelled "active-active" conflict scenario.

## Policy Implications

1. **Central bank independence:** Quantify the fiscal conditions under which
   de facto monetary autonomy is compromised, informing institutional design.
2. **Debt sustainability:** Characterise the nonlinear relationship between debt
   levels and risk premia, with implications for fiscal rules.
3. **Financial stability:** Model how sovereign stress propagates to private credit
   via the banking sector, informing macroprudential policy.
4. **International spillovers:** Assess how domestic fiscal-monetary tensions
   affect capital flows and exchange rates, with implications for
   international policy coordination.
